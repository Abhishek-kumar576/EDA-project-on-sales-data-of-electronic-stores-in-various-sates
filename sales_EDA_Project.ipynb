```python
# handling missng values with the mean
mean_profit = a['profit'].mean()
a['profit'] = a['profit'].fillna(mean_profit)
```

```python
import matplotlib.pyplot as plt
plt.scatter(a['profit'],a['sales'])
plt.xlabel('profit')
plt.ylabel('sales')
plt.title('scatter plot')
```

```python
plt.hist(a['profit'], bins=5)
plt.xlabel('profit')
plt.ylabel('Frequency')
plt.title('histogram of profit')
```

```python
plt.plot(a['region'],a['state'])
plt.xlabel('region')
plt.ylabel('state')
plt.title('line plot region by state')
```

```python
plt.boxplot(a['profit'])

```

```python
profit_by_state = a.groupby('state')['profit'].sum()
plt.pie(profit_by_state, labels=profit_by_state.index, autopct='%1.1f%%')
plt.title('Total Profit by State')
plt.axis('equal')
```
